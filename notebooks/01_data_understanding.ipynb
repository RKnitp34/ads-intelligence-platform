### Business Question

```python
### Q1. Which users are the most active?
(
    train
    .groupby("user_id")
    .size()
    .sort_values(ascending=False)
    .head(20)
)
```

```python
### Q2. Which videos are recommended the most?
(
    train
    .groupby("video_id")
    .size()
    .sort_values(ascending=False)
    .head(20)

)
```

```python
### Q3. Which days have the highest traffic?

(
    train
    .groupby("date")
    .size()
    .sort_values(ascending=False)
)
```

```python
### Q4. Which hours are busiest?

train["hour"] = (
    train["hourmin"] // 100
)

(
    train
    .groupby("hour")
    .size()
    .sort_values(ascending=False)
)
```
